## Data Loading

### Accuracy Data Loading

```python
import numpy as np

# load the baseline cnn accuralty results
acc_results1 = np.loadtxt("results/base_cnn/acc_base_cnn.txt")

#show the number of rows and columns
print(acc_results1.shape)
```

### Loss Data Loading

```python
#load the baseline cnn loss results
loss_results1=np.loadtxt("results/base_cnn/loss_base_cnn.txt")

#show the number of rows and columns
print(loss_results1.shape)
```

_____

## Accuracy Graph

```python
import matplotlib.pyplot as plt

#plot each row of accuracy resutls
plt.figure(figsize=(10, 6))
plt.plot(acc_results1, label="Base Model")

plt.title("Baseline vs. Reinforced CNN Validation Accuracy")
plt.xlabel("Validation Step")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

----

## Loss Graph

```python
import matplotlib.pyplot as plt

#plot each row of accuracy resutls
plt.figure(figsize=(10, 6))
plt.plot(loss_results1, label="Base Model")

plt.title('Baseline CNN val Loss Results')
plt.xlabel('Baseline CNN validation step')
plt.xlabel("Validation Step")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
