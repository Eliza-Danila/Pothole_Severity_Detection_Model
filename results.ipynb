## Accuracy Graphs

### Base CNN

```python
import matplotlib.pyplot as plt

#plot each row of accuracy resutls
for i, row in enumerate(acc_results1):
    plt.plot(row, label=f'Run {i+1}')
plt.title('Baseline CNN Accuracy Results')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
